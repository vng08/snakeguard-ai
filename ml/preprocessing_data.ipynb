```python
testing_df = pd.read_csv("../data/raw/SnakeCLEF2023-PubTestMetadata.csv")

print("Test set information")
print("Shape:", testing_df.shape)
testing_df.head()
```

## 2. Extract snake species found in Vietnam from The Reptile Database. (Extract via HTML)

```python
with open("../data/raw/ReptileDatabase-VNSpecies.html", "r", encoding="utf-8") as f:
    source_page = BeautifulSoup(f, "html.parser")

raw_html = "\n".join(line.get_text() for line in source_page.select(".line-content"))
soup = BeautifulSoup(html.unescape(raw_html), "html.parser")
species_names = [tag.get_text(strip=True) for tag in soup.select("li a em")]
vietnam_snakes = pd.DataFrame({"scientific_name": species_names})

print("Shape:", vietnam_snakes.shape)
vietnam_snakes.head()
```

```python
vietnam_snakes.to_csv("../data/raw/ReptileDatabase-VNSpecies.csv", index=False)
```

## 3. Concatenate the training dataset with the additional dataset to create a large training dataset.

```python
combined_training_df = pd.concat([training_df, additional_training_df], ignore_index=True)

print("Combined training set")
print("Shape:", combined_training_df.shape)
print("Species:", combined_training_df["binomial_name"].nunique())
print("Observations:", combined_training_df["observation_id"].nunique())

combined_training_df.head()
```

## 4. Filter Vietnam species

```python
# Checking class id vs binomial_name
name_count = combined_training_df["binomial_name"].nunique()
class_count = combined_training_df["class_id"].nunique()

print("Scientific names:", name_count)
print("Class IDs:", class_count)
print("Same number:", name_count == class_count)
print("Each name maps to one class ID:", combined_training_df.groupby("binomial_name")["class_id"].nunique().max() == 1)
print("Each class ID maps to one name:",combined_training_df.groupby("class_id")["binomial_name"].nunique().max() == 1)
```

```python
vietnam_species = set(vietnam_snakes["scientific_name"])
snakeclef_species = set(combined_training_df["binomial_name"])
matched_species = vietnam_species & snakeclef_species

print("Vietnam species master:", len(vietnam_species))
print("SnakeCLEF species:", len(snakeclef_species))
print("Matched Vietnam species:", len(matched_species))

vietnam_training_df = combined_training_df[combined_training_df["binomial_name"].isin(matched_species)].copy()

print("Images:", len(vietnam_training_df))
print("Observations:", vietnam_training_df["observation_id"].nunique())
vietnam_training_df.head()
```

## 5. Analyze species

```python
species_counts = (vietnam_training_df.groupby("binomial_name").size().sort_values(ascending=False))

print("Matched Vietnam species:", species_counts.shape[0])
print(species_counts.describe())

print("\n>= 500 images:", (species_counts >= 500).sum())
print(">= 250 images:", (species_counts >= 200).sum())
print(">= 100 images:", (species_counts >= 100).sum())
print(">= 50 images:", (species_counts >= 50).sum())
print("< 50 images:", (species_counts < 50).sum())

species_counts.head()
```

## 6. Add medically important venomous snake (MIVS) & select final classes

```python
mivs_df = pd.read_csv("http://ptak.felk.cvut.cz/plants/plants/SnakeCLEF2023/venomous_status_list.csv")
mask = mivs_df["MIVS"] == 1

print("Shape: ", mivs_df.shape)
print("Has venomous: ", mask.sum())
mivs_df.head()
```

```python
candidate_stats = species_counts.reset_index()
candidate_stats.columns = ["binomial_name", "image_count"]

class_map = vietnam_training_df[["binomial_name", "class_id"]].drop_duplicates()

candidate_stats = candidate_stats.merge(class_map, on="binomial_name",how="left")
candidate_stats = candidate_stats.merge(mivs_df, on="class_id", how="left")

candidate_stats = candidate_stats[["binomial_name", "image_count", "class_id", "MIVS"]]

print(candidate_stats["MIVS"].value_counts(dropna=False))
candidate_stats.head()
```

```python
# select final classes
final_classes = candidate_stats[(candidate_stats["MIVS"] == 1) |(candidate_stats["image_count"] >= 50)].copy()
final_classes = final_classes.sort_values(["MIVS", "image_count"], ascending=[False, False]).reset_index(drop=True)

print("Final classes:", len(final_classes))
print("MIVS classes:", (final_classes["MIVS"] == 1).sum())
print("Non-MIVS classes:", (final_classes["MIVS"] == 0).sum())

final_classes.describe()
```

```python
print(final_classes.isna().sum())
final_classes.head()
```

## 7. Build class mapping

```python
class_mapping = (final_classes.sort_values("class_id").reset_index(drop=True).copy())
class_mapping["label_idx"] = range(len(class_mapping))
class_mapping = class_mapping[["label_idx", "class_id", "binomial_name", "MIVS", "image_count"]]

print("Number of classes:", len(class_mapping))
class_mapping.head()
```

```python
print("Unique labels:", class_mapping["label_idx"].nunique())
print("Unique class IDs:", class_mapping["class_id"].nunique())
print("Unique species:", class_mapping["binomial_name"].nunique())

assert class_mapping["label_idx"].tolist() == list(range(len(class_mapping)))
```

## 8. Filter for validation set

```python
final_class_ids = set(final_classes["class_id"])
filtered_validation_df = validation_df[validation_df["class_id"].isin(final_class_ids)].copy()

print("Validation images:", len(filtered_validation_df))
print("Validation species:", filtered_validation_df["class_id"].nunique())
print("Missing values:")
print(filtered_validation_df.isna().sum())

filtered_validation_df.head()
```

## 9. Prepare Metadata

### Why We Do Not Use the Official Public Test Set

The official SnakeCLEF 2023 Public Test metadata does not provide ground-truth labels such as `class_id` or `binomial_name`. Therefore, it cannot be used to calculate classification metrics such as accuracy, F1-score, or Top-K accuracy.

For this project, the original training data is split by `observation_id` into training and validation sets to avoid data leakage between images from the same observation. The official SnakeCLEF validation set is then used as the final labeled test set for model evaluation.

This results in the following evaluation setup:

- **Train set:** model training
- **Validation set:** hyperparameter tuning and early stopping
- **Test set:** final evaluation using the official SnakeCLEF validation data

```python
final_class_ids = set(final_classes["class_id"])
full_train_metadata = vietnam_training_df[vietnam_training_df["class_id"].isin(final_class_ids)].copy()
full_train_metadata = full_train_metadata.merge(class_mapping[["class_id", "label_idx", "MIVS"]], on="class_id", how="left")

print("Images:", len(full_train_metadata))
print("Species:", full_train_metadata["class_id"].nunique())
print("Observations:", full_train_metadata["observation_id"].nunique())
print("Has missing values:", full_train_metadata.isna().any().any())
```

```python
observation_class_count = (full_train_metadata.groupby("observation_id")["class_id"].nunique())
print("Observations mapped to >1 class:", (observation_class_count > 1).sum())

observation_df = (full_train_metadata[["observation_id", "class_id"]].drop_duplicates())
observation_counts = observation_df["class_id"].value_counts()
print("Minimum observations per class:", observation_counts.min())
```

```python
train_observations, val_observations = train_test_split(observation_df, test_size=0.1, random_state=42, stratify=observation_df["class_id"])
train_metadata = full_train_metadata[full_train_metadata["observation_id"].isin(train_observations["observation_id"])].copy()
val_metadata = full_train_metadata[full_train_metadata["observation_id"].isin(val_observations["observation_id"])].copy()

print("Train images:", len(train_metadata))
print("Train observations:", train_metadata["observation_id"].nunique())
print("Train species:", train_metadata["class_id"].nunique())

print("\nValidation images:", len(val_metadata))
print("Validation observations:", val_metadata["observation_id"].nunique())
print("Validation species:", val_metadata["class_id"].nunique())

overlap = set(train_metadata["observation_id"]) & set(val_metadata["observation_id"])
print("Train/Val observation overlap:", len(overlap))
```

```python
test_metadata = filtered_validation_df.copy()
test_metadata = test_metadata.merge(class_mapping[["class_id", "label_idx", "MIVS"]], on="class_id", how="left")
test_metadata = test_metadata[
    [
        "observation_id",
        "binomial_name",
        "class_id",
        "label_idx",
        "image_path",
        "MIVS"
    ]
]

print("Test images:", len(test_metadata))
print("Test species:", test_metadata["class_id"].nunique())
print("Missing values:", test_metadata.isna().any().any())

test_metadata.head()
```

## 10. Final checks & Save 

```python
final_class_ids = set(class_mapping["class_id"])

print("Train:", len(train_metadata), "images |", train_metadata["class_id"].nunique(), "classes")
print("Val:", len(val_metadata), "images |", val_metadata["class_id"].nunique(), "classes")
print("Test:", len(test_metadata), "images |", test_metadata["class_id"].nunique(), "classes")
print("Final classes:", len(class_mapping))

print("\nMissing values:")
print("Train:", train_metadata.isna().any().any())
print("Val:", val_metadata.isna().any().any())
print("Test:", test_metadata.isna().any().any())

print("\nInvalid class IDs:")
print("Train:", len(set(train_metadata["class_id"]) - final_class_ids))
print("Val:", len(set(val_metadata["class_id"]) - final_class_ids))
print("Test:", len(set(test_metadata["class_id"]) - final_class_ids))

train_val_overlap = set(train_metadata["observation_id"]) & set(val_metadata["observation_id"])
print("\nTrain/Val observation overlap:", len(train_val_overlap))

print("\nDuplicate image_path within each split:")

print("Train:", train_metadata["image_path"].duplicated().sum())
print("Val:", val_metadata["image_path"].duplicated().sum())
print("Test:", test_metadata["image_path"].duplicated().sum())
```

```python
processed_dir = Path("../data/processed")
processed_dir.mkdir(parents=True, exist_ok=True)

train_metadata.to_csv(processed_dir / "train.csv", index=False)
val_metadata.to_csv(processed_dir / "val.csv", index=False)
test_metadata.to_csv(processed_dir / "test.csv", index=False)
class_mapping.to_csv(processed_dir / "classes.csv", index=False)
final_classes.to_csv(processed_dir / "species_stats.csv", index=False)

print("Processed metadata saved.")
```
